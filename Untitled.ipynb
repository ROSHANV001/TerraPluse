```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from terrapulse_ai import TerraPulseAI, IDEAL_CROP_PROFILES
```

```python
df = pd.read_csv("terrapulse_dataset_35k.csv")
print(df.shape)
df.head(10)          # shows as a neat table automatically
```

```python
df['crop'].value_counts().plot(kind='bar', color='steelblue')
plt.title("Records per crop")
plt.tight_layout()
plt.show()
```

```python
ai = TerraPulseAI()
ai.train(df)         # accuracy report prints right here in the cell
```

```python
import pickle

with open("terrapulse_recommender.pkl", "wb") as f:
    pickle.dump(ai.recommender, f)

with open("terrapulse_predictor.pkl", "wb") as f:
    pickle.dump(ai.predictor, f)

print("Models saved.")
```

```python
from terrapulse_ai import SensorReading

reading = SensorReading(
    zone_id=1, N=55, P=40,
    K=38, pH=5.9, moisture=42, temperature=23
)
result = ai.run(reading)
result   # prints as clean dict in notebook
```
